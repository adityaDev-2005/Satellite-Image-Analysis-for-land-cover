Implementation

```python
from torch.utils.data import Dataset
import pandas as pd
from torchvision.io import decode_image
import os

class EuroSATDataset(Dataset):
  def __init__(self,annotations_file,img_dir,transform=None,target_transform=None):
    self.img_labels = pd.read_csv(annotations_file)
    self.img_dir = img_dir
    self.transform = transform
    self.target_transform = target_transform

  def __len__(self):
    return len(self.img_labels)


  def __getitem__(self,idx):
    img_path = os.path.join(self.img_dir, self.img_labels.iloc[idx,1])
    image = decode_image(img_path)
    label = self.img_labels.iloc[idx,2]
    if self.transform:
      image = self.transform(image)
    if self.target_transform:
      label = self.target_transform(label)
    return image,label

```

Making of the CNN model

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
```

```python
class EuroSATCNN(nn.Module):
  def __init__(self):
    super(EuroSATCNN,self).__init__()

    self.conv1 = nn.Conv2d(in_channels=3,out_channels=16,kernel_size=3)
    self.conv2 = nn.Conv2d(in_channels=16,out_channels=32,kernel_size=3)
    self.activation = nn.ReLU()

    # first fully-connected layer
    self.fc1 = nn.Linear(in_features=6272,out_features=128)

    # final classfication layer
    self.fc2 = nn.Linear(in_features=128,out_features=10)

  def forward(self,x):
    # 1st convolution
    x = self.conv1(x)
    x = self.activation(x)

    # pooling
    x = F.max_pool2d(x,kernel_size=(2,2))

    # 2nd convolution
    x = self.conv2(x)
    x = self.activation(x)

    # pooling
    x = F.max_pool2d(x,kernel_size=(2,2))

    x = torch.flatten(x,start_dim=1)

    # fully conencted layer
    x = self.fc1(x)

    x = self.activation(x)

    x = self.fc2(x)

    return x

model = EuroSATCNN()
print(model)

params = list(model.parameters())


x = torch.randn(32, 3, 64, 64)
output = model(x)
print(output.shape)



# x = torch.randn(32, 3, 64, 64)
# print(x.shape)
```

Training the CNN (currently for 32 images)

```python
images, labels = next(iter(train_loader))

print(images.shape)
print(labels.shape)
print(labels.dtype)
print(labels.min(), labels.max())
```

```python
images, labels = next(iter(train_loader))

criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)

model.train()
optimizer.zero_grad()

outputs = model(images)

loss = criterion(outputs,labels)
loss.backward()
optimizer.step()
print("Loss:", loss.item())

predicted = outputs.argmax(dim=1)
print("Predicted Classes : ",predicted)

correct = (predicted == labels).sum().item()

accuracy = (correct/len(labels))*100

print("Correct :",correct)
print("Accuracy : ",accuracy)
```

Main training loop

```python
import torch

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("Using device:", device)
```

```python
from torch.utils.data import DataLoader

val_dataset = EuroSATDataset(
    "/kaggle/input/eurosat-dataset/EuroSAT/validation.csv",
    "/kaggle/input/eurosat-dataset/EuroSAT",
    transform=transform
)

val_loader = DataLoader(
    val_dataset,
    batch_size=32,
    shuffle=False,
    num_workers=0
)

print("Validation images:", len(val_dataset))
print("Validation batches:", len(val_loader))
```

```python
epochs = 10

for epoch in range(epochs):
    # Training
    model.train()

    running_loss = 0.0
    correct = 0
    total = 0

    for images, labels in train_loader:
        images = images.to(device)
        labels = labels.to(device)

        optimizer.zero_grad()

        outputs = model(images)
        loss = criterion(outputs, labels)

        loss.backward()
        optimizer.step()

        # Accumulate loss weighted by batch size
        batch_size = labels.size(0)
        running_loss += loss.item() * batch_size

        predicted = outputs.argmax(dim=1)
        correct += (predicted == labels).sum().item()
        total += batch_size

    train_loss = running_loss / total
    train_accuracy = 100 * correct / total

    # Validation
    model.eval()

    val_loss_total = 0.0
    val_correct = 0
    val_total = 0

    with torch.no_grad():
        for images, labels in val_loader:
            images = images.to(device)
            labels = labels.to(device)

            outputs = model(images)
            loss = criterion(outputs, labels)

            batch_size = labels.size(0)
            val_loss_total += loss.item() * batch_size

            predicted = outputs.argmax(dim=1)
            val_correct += (predicted == labels).sum().item()
            val_total += batch_size

    val_loss = val_loss_total / val_total
    val_accuracy = 100 * val_correct / val_total

    print(
        f"Epoch [{epoch + 1}/{epochs}] | "
        f"Train Loss: {train_loss:.4f} | "
        f"Train Acc: {train_accuracy:.2f}% | "
        f"Val Loss: {val_loss:.4f} | "
        f"Val Acc: {val_accuracy:.2f}%"
    )
```
